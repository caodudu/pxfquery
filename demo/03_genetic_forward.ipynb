# Genetic + Forward Query

Potential Fig5-style scene: inspect functional effects after a genetic perturbation.

This demo uses the current PxFquery public interface: a biomedical question goes through `PxFQuery().ask(...)` and returns a human-readable answer with evidence and limitations.

Question:

```text
What happens to functional programs if I knock down KRAS in A549 lung cancer cells?
```


````python
from pxfquery import PxFQuery

pxf = PxFQuery()
answer = pxf.ask('What happens to functional programs if I knock down KRAS in A549 lung cancer cells?')

print(answer)
````

## Structured View

The structured view is secondary, but useful for tests and reproducibility.

````python
from pprint import pprint

pprint({
    "interpreted_question": answer.interpreted_question,
    "top_results": answer.biological_results[:5],
    "evidence": answer.evidence,
    "limitations": answer.limitations,
})
````
